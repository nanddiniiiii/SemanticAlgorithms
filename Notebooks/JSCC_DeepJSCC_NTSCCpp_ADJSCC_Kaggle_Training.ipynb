# DeepJSCC, NTSCC++, and ADJSCC: completed Kaggle comparison

All three models completed 150 epochs on AWGN and 150 epochs on Rayleigh at 10 dB; each evaluated all 24 Kodak images. The NTSCC++-named experiment uses the author-linked `NTSCC_plus` implementation. Training jobs ran in three private Kaggle notebooks because the official models use separate framework/runtime dependencies. This notebook consolidates their completed results; the linked kernels contain the executed epoch records and outputs.

## Training protocol and comparison limits

- DeepJSCC and ADJSCC each sampled all 800 DIV2K images per epoch with 128×128 crops, for 150 full passes; batch sizes 32 and 16.
- NTSCC++ sampled 128 distinct images per epoch from the 800-image pool, 256×256 crops, batch 8, FP16 autocast: 19,200 sampled crops across training (about 24 image exposures/image), not 150 full passes.
- DeepJSCC and ADJSCC nominal CBR is 1/6. NTSCC++ averaged 0.047–0.049 CBR; reported quality is not a matched-rate comparison.
- The NTSCC_plus release supports AWGN. Its Rayleigh result extends the implementation with quasi-static complex fading and perfect receiver CSI/equalization.
